```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv('/content/drive/MyDrive/Colab Notebooks/M L training Winnovation /Transformed Data Set - Sheet1.csv')
df.head(11)
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df.columns
```

```python
#Inplace encoding the values of the column.
cols_to_encode = [
                  'Favorite Color',
    'Favorite Music Genre',
    'Favorite Beverage',
    'Favorite Soft Drink',
]

for col in cols_to_encode:
  df[col], _ = pd.factorize(df[col])

# Verify the changes
df.head()
```

```python
#Encoding the Gender column
df['Male_Count'] = np.where(df['Gender']=='M', 1, 0)
df['Female_Count'] = np.where(df['Gender']=='F', 1, 0)
```

```python
df.sample(11)
```

```python
x=df.drop(columns= ['Gender'])
y=df['Gender']
```

```python
y, _ = pd.factorize(y)
```

```python
#Spliting data into Train and Test
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2, random_state = 42)
```

```python
#Fitting Logistic Regression in Modal
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
train_pred = model.predict(x_train)
test_pred = model.predict(x_test)
```

```python
train_pred[:5]
```

```python
test_pred[:5]
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, recall_score, f1_score
```

```python
print('Train accuracy_score :-',accuracy_score(y_train,train_pred))
print('Train precision_score :-',precision_score(y_train,train_pred))
print('Train recall_score :-',recall_score(y_train,train_pred))
print('Train confusion_matrix :-\n',confusion_matrix(y_train,train_pred))
```

```python
print('Test accuracy_score :-',accuracy_score(y_test,test_pred))
print('Test precision_score :-',precision_score(y_test,test_pred))
print('Test recall_score :-',recall_score(y_test,test_pred))
print('Test confusion_matrix :-\n',confusion_matrix(y_test,test_pred))
```
