```python
x = df.drop(columns='Gender', axis=1)
y = df['Gender']
```

```python
x = pd.get_dummies(x, drop_first=True, dtype=int)
x.sample(11)
```

```python
y = y.map({'F': 0, 'M': 1})
y.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=2, stratify=y)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
import tensorflow as tf
tf.random.set_seed(3)
from tensorflow import keras
```

```python
model = keras.Sequential([
    keras.layers.Dense(30, input_shape=(16,), activation='relu'),
    keras.layers.Dense(15, activation='relu'),
    keras.layers.Dense(1, activation='sigmoid')
])
```

```python
model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(x_train,y_train,validation_split=0.1,epochs=15)
```

```python
train_pred = model.predict(x_train)
train_pred[:5]
```

```python
train_pred = np.argmax(train_pred,axis=1)
train_pred[:5]
```

```python
test_pred = model.predict(x_test)
test_pred = np.argmax(test_pred,axis=1)
test_pred[:5]
```

```python
from sklearn.metrics import accuracy_score
```

```python
print("Training Accuracy:", accuracy_score(y_train, train_pred))
print("Testing Accuracy:", accuracy_score(y_test, test_pred))
```

```python
import gradio as gr
import numpy as np

# Creating the prediction function

def predict_gender(color, music, beverage, soft_drink):

    # Converting user inputs into a DataFrame
    input_data = pd.DataFrame({
        'Favorite Color': [color],
        'Favorite Music Genre': [music],
        'Favorite Beverage': [beverage],
        'Favorite Soft Drink': [soft_drink]
    })

    # Applying One-Hot Encoding
    input_data = pd.get_dummies(input_data, drop_first=True, dtype=int)

    # Matching the training columns
    input_data = input_data.reindex(columns=x.columns, fill_value=0)

    # Standardizing the input data
    input_data_scaled = scaler.transform(input_data)

    # Making predictions
    prediction = model.predict(input_data_scaled, verbose=0)

    # Extracting prediction probability
    probability = prediction[0][0]

    # Returning confidence scores
    return {
        "Female": float(1 - probability),
        "Male": float(probability)
    }


# Creating input fields for the web application

inputs = [

    gr.Dropdown(choices=df['Favorite Color'].unique().tolist(), label='Favorite Color'),

    gr.Dropdown(choices=df['Favorite Music Genre'].unique().tolist(), label='Favorite Music Genre'),

    gr.Dropdown(choices=df['Favorite Beverage'].unique().tolist(), label='Favorite Beverage'),

    gr.Dropdown(choices=df['Favorite Soft Drink'].unique().tolist(), label='Favorite Soft Drink')

]

# Defining the Gradio interface

interface = gr.Interface(

    fn=predict_gender,

    inputs=inputs,

    outputs=gr.Label(num_top_classes=2),

    title="Gender Prediction App",

    description="Select your favorite preferences below to predict Gender using our trained Artificial Neural Network."

)

# Launching the web application

interface.launch(share=True)
```
