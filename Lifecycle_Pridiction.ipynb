```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('/content/drive/MyDrive/CV Data Life Expectancy Data.csv')
df.head()
```

```python
df.info()
```

```python
df['Status'].value_counts()
```

```python
df['Developing_Status'] = np.where(df['Status'] == 'Developing',1,0)
df['Developed_Status'] = np.where(df['Status'] == 'Developed',1,0)
```

```python
df.info()
```

```python
df['Country_Encoded'] = df['Country'].astype('category').cat.codes
#df['Country_Encoded'] = pd.factorize(df['Country'])[0]
```

```python
df.sample(11)
```

```python
df = df.drop(['Country'],axis=1)
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df['Life expectancy '].fillna(df['Life expectancy '].median(), inplace=True)
df['Adult Mortality'].fillna(df['Adult Mortality'].median(), inplace=True)
df['Hepatitis B'].fillna(df['Hepatitis B'].median(), inplace=True)
df['Measles '].fillna(df['Measles '].median(), inplace=True)
df[' BMI '].fillna(df[' BMI '].median(), inplace=True)
df['under-five deaths '].fillna(df['under-five deaths '].median(), inplace=True)
df[' HIV/AIDS'].fillna(df[' HIV/AIDS'].median(), inplace=True)
df['Total expenditure'].fillna(df['Total expenditure'].median(), inplace=True)
df['GDP'].fillna(df['GDP'].median(), inplace=True)
df['Population'].fillna(df['Population'].median(), inplace=True)
df[' thinness  1-19 years'].fillna(df[' thinness  1-19 years'].median(), inplace=True)
df[' thinness 5-9 years'].fillna(df[' thinness 5-9 years'].median(), inplace=True)
df['Income composition of resources'].fillna(df['Income composition of resources'].median(), inplace=True)
df['Schooling'].fillna(df['Schooling'].median(), inplace=True)
df['Alcohol'].fillna(df['Alcohol'].median(), inplace=True)
df['Polio'].fillna(df['Polio'].median(), inplace=True)
df['Diphtheria '].fillna(df['Diphtheria '].median(), inplace=True)
```

```python
df.isnull().sum()
```

```python
x = df.drop(['Life expectancy ','Status'],axis=1)
y = df['Life expectancy ']
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error,mean_absolute_percentage_error
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
model.coef_
```

```python
train_pred = model.predict(x_train)
test_pred = model.predict(x_test)
```

```python
train_pred[:5]
```

```python
y_train.values[:5]
```

```python
print('Train  mean_squared_error :-',mean_squared_error(y_train,train_pred))
print('Train  mean_absolute_error :-',mean_absolute_error(y_train,train_pred))
print('Train  mean_absolute_percentage_error :-',mean_absolute_percentage_error(y_train,train_pred)*100)
```

```python
print('Test mean_squared_error :-',mean_squared_error(y_test,test_pred))
print('Test mean_absolute_error :-',mean_absolute_error(y_test,test_pred))
print('Test mean_absolute_percentage_error :-',mean_absolute_percentage_error(y_test,test_pred)*100)
```
